Creating a function to automatically average a numerical column that is grouped by a string column.

```python
##Function 1
def average_by_group(df, group_col, value_col):
        return df.groupby(group_col)[value_col].mean().sort_values(ascending=False)
```

```python
average_by_group(df, "gender","stress_level")
```

Average stress level for all 3 gender categories are extremely close. Averages are within plus or minus 0.2. Would like to add min and max to the function as well.

```python
##Function 1
def summary_by_group(df, group_col, value_col):
        return df.groupby(group_col)[value_col].agg(['min','mean', 'max'])
```

```python
average_by_group(df, "occupation", "daily_screen_time_hours")
```

Create a function that will make a column categorizing the screen time into low, moderate, and high groups based off of the numerical value of daily_screen_time_hours. 

```python
##function 2
def screen_time_category(hours):
        if hours < 3:
                return "Low"
        elif hours < 6:
                return "Moderate"
        else:
                return "High"
```

```python
df['screen_time_category'] = df['daily_screen_time_hours'].apply(screen_time_category)
```
