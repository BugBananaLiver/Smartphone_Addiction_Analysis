Creating function to auto average a numerical column grouped by a string column

```python
def average_by_group(df, group_col, value_col):
        return df.groupby(group_col)[value_col].mean().sort_values(ascending=False)

```

```python
average_by_group(df, "gender","stress_level")
```

```python
average_by_group(df, "occupation", "daily_screen_time_hours")
```

```python
def screen_time_category(hours):
        if hours < 3:
                return "Low"
        elif hours < 6:
                return "Moderate"
        else:
                return "High"
```

```python
df['screen_time_category'] = df['daily_screen_time_hours'].apply(screen_time_category)
```
