```python
#What is the average screen time based on gender for d1?
with sqlite3.connect('Smartphone.db') as connection:
    cursor = connection.cursor()

    cursor.execute("""
        SELECT u.gender, AVG(h.daily_screen_time_hours) AS avg_hours
        FROM users_d1 u
        JOIN hours_d1 h on u.user_id = h.user_id
        GROUP BY u.gender;
        """)

    rows = cursor.fetchall()

    for row in rows:
        print(row)
```

```python
#What is the average number of notifications per day by gender in d1?

with sqlite3.connect('Smartphone.db') as connection:
    cursor = connection.cursor()

    cursor.execute("""
        SELECT u.gender, AVG(b.notifications_per_day)
        FROM users_d1 u
        JOIN behavior_d1 b 
        ON u.user_id = b.user_id
        GROUP BY u.gender;
        """)

    rows = cursor.fetchall()

    for row in rows:
        print(row)
```
