    def summary_by_group(df, group_col, value_col):

    ```
    def screen_time_category(hours):

                return "Low"
            elif hours < 6:
                return "Moderate"
            else:
                return "High"

    ```

````python
#Which occupations have the highest average daily screen time?

with sqlite3.connect('Smartphone.db') as connection:
    cursor = connection.cursor()

    cursor.execute("""
        SELECT u.occupation, AVG(h.daily_screen_time_hours) AS average_screen_time
        FROM users_d2 u JOIN hours_d2 h ON u.user_id = h.user_id
        GROUP BY u.occupation
        ORDER BY average_screen_time DESC;        
                """)

    rows = cursor.fetchall()

occupation = [row[0] for row in rows]
average_screen_time = [row[1] for row in rows]

plt.figure(figsize=(10,6))

plt.barh(occupation, average_screen_time,color='steelblue')

plt.title(
    "Average Daily Screen Time by Occupation",
    fontsize=16,
    fontweight='bold',
    color='darkblue'
)

plt.xlabel(
    "Average Daily Screen Time (Hours)",
    fontsize=12,
    fontweight='bold',
    color='darkred'
)

plt.ylabel(
    "Occupation",
    fontsize=12,
    fontweight='bold',
    color='darkred'
)

plt.xticks(
    fontsize=10,
    fontweight='bold')
plt.yticks(
    fontsize=10,
    fontweight='bold'
)

plt.gca().invert_yaxis()
#puts highest value at the top

for i, value in enumerate(average_screen_time):
    plt.text(value + 0.01, i, f"{value:.2f}",
             va='center',
             fontweight='bold')
#adds value to end of bars

plt.grid(axis='x', linestyle='--', alpha=0.5)
#light grid

ax = plt.gca()

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.show()
    # for row in rows:
    #     print(row)
````

````python
#How many users spend more than the average daily screen time?

with sqlite3.connect('Smartphone.db') as connection:
    cursor = connection.cursor()

    cursor.execute("""
        SELECT COUNT(u.user_id)
        FROM users_d1 u JOIN hours_d1 h ON u.user_id = h.user_id
        WHERE h.daily_screen_time_hours >
            (SELECT AVG(h.daily_screen_time_hours)
            FROM hours_d1 h)
        ORDER BY u.user_id;        
                """)

    rows = cursor.fetchall()

    for row in rows:
        print(row)
````

````python
with sqlite3.connect('Smartphone.db') as connection:
    cursor = connection.cursor()

    cursor.execute("""
        SELECT 
            CASE
                WHEN h.daily_screen_time_hours < 4 THEN 'Low'
                WHEN h.daily_screen_time_hours > 6 THEN 'High'
                ELSE 'moderate'
            END AS screen_time_group,
            AVG(h.sleep_hours) AS avg_sleep
        FROM hours_d1 h
        GROUP BY screen_time_group
                """)

    rows = cursor.fetchall()

    for row in rows:
        print(row)
````

````python
#Do occupations with higher screen time also receive more notifications per day?

with sqlite3.connect('Smartphone.db') as connection:
    cursor = connection.cursor()

    cursor.execute("""
        SELECT u.occupation,
            AVG(h.daily_screen_time_hours) AS avg_screen_time,
            AVG(w.notifications_received_per_day) AS avg_notifications
        FROM users_d2 u
        JOIN hours_d2 h ON u.user_id = h.user_id
        JOIN wellbeing_d2 w ON u.user_id = w.user_id
        GROUP BY u.occupation
        ORDER BY avg_screen_time DESC
                """)

    rows = cursor.fetchall()

occupation = [row[0] for row in rows]
avg_screen_time = [row[1] for row in rows]
avg_notifications = [row[2] for row in rows]

plt.figure(figsize=(8,6))
plt.scatter(avg_notifications, avg_screen_time)

plt.xlabel('Average Notifications per Day', fontsize=12, fontweight='bold', color='darkred')
plt.ylabel('Average Daily Screen Time (Hours)', fontsize=12, fontweight='bold', color='darkred')
plt.title('Average Notifications vs Average Screen Time by Occupation', fontsize=12, fontweight='bold', color='darkblue')

for i in range(len(occupation)):
    plt.text(
        avg_notifications[i] + 0.05,
        avg_screen_time[i] + 0.002,
        occupation[i],
        fontsize=8,
        fontweight= 'bold',
        color='navy',
    )

    plt.scatter(
        avg_notifications,
        avg_screen_time,
        color='steelblue',
        s=80
    )

plt.grid(True)
plt.show()

    # for row in rows:
    #     print(row)
````
